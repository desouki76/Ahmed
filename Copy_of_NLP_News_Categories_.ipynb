```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python

# Step 1: Convert text into numeric vectors (TF-IDF)

vectorizer = TfidfVectorizer(max_features=5000)  # limit features
X_tfidf = vectorizer.fit_transform(X).toarray()
```

```python
#split
#split Data
X=new_df['headline']
y=new_df['category_encoded']
#split data
X_train, X_test, y_train, y_test = train_test_split(X_tfidf, y, test_size=0.2, random_state=42, stratify=new_df.category_encoded)
```

```python
# Step 3: Apply SMOTE
smote = SMOTE(random_state=42, sampling_strategy="auto")
X_resampled, y_resampled = smote.fit_resample(X_train, y_train)

print("Resampled training set distribution:", Counter(y_resampled))
```
