```python
# Initialize tools
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
```

```python
# Preprocessing function
def preprocess(text):
    tokens = nltk.word_tokenize(text.lower())  # tokenize + lowercase
    tokens = [t for t in tokens if t.isalpha()]  # keep only words
    tokens = [t for t in tokens if t not in stop_words]  # remove stopwords
    tokens = [lemmatizer.lemmatize(t) for t in tokens]  # lemmatization
    return " ".join(tokens)
```

```python

# Apply preprocessing
clean_corpus = [preprocess(doc) for doc in corpus]

print("Cleaned Corpus:", clean_corpus)
```

```python

# Apply N-grams (1 to 3 → unigram, bigram, trigram)
vectorizer = CountVectorizer(ngram_range=(1,3))
X = vectorizer.fit_transform(clean_corpus)

print("\nVocabulary:", vectorizer.get_feature_names_out())
print("\nMatrix:\n", X.toarray())
```
