```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from skimage.io import imread, imsave, imshow
```

```python
img_camera= imread('/content/CamMan.jpg')
img_camera.shape
imshow(img_camera)
```

#Cordiate transformation


```python
from skimage import transform
```

```python
#tfrom = transform.SimilarityTransform(translation=(100,100 ))

tfrom = transform.SimilarityTransform(matrix = np.array([[1,0,100],[0,1,100],[0,0,1]]))

tfrom
```

```python
transformed_camera = transform.warp(img_camera, tfrom)
imshow(transformed_camera)
```

```python
#shift in the اتجاه العكسى
transformed_camera =  transform.warp(img_camera, tfrom.inverse)
imshow(transformed_camera)
```
