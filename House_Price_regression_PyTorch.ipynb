# معالجة Missing Data

```python
null_ratio = df.isnull().sum() / len(df)
# drop coulmm if the mnull data ration > 30%
df = df.drop(columns = null_ratio[null_ratio > 0.3].index, axis=1)
#df=df.drop(null_ratio[null_ratio > 0.3].index, axis=1)
```

if i have missing data from kind object or string i will fill with imunit and if i have data missing object i will fill it with mode

```python
#string missing will be filled by mode and string will be filled with Mode
# لف على مكل ال العواميد و شيك على توعيه داتا
for col in df.columns:
  if df[col].dtype == 'object':
    df[col] = df[col].fillna(df[col].mode()[0])
  else:
    df[col] = df[col].fillna(df[col].mean())
#
```
